```python
def train_model(model,*data_items,**keys):
    print(f"version:{model}")
    print("\ndata items:")
    for item in data_items:
        print(f"data item: {item}")
    print("\nkeys:")
    for key, value in keys.items():
        print(f"{key}: {value}")

train_model("v1.0", "data1", "data2", "data3", learning_rate=0.01, batch_size=32)
```
