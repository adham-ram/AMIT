```python
def factorial(num: int) -> int:
    """Calculates the factorial of a non-negative integer using recursion.

    Args:
        num (int): A non-negative integer whose factorial is to be calculated.

    Returns:
        int: The factorial of the given number. Returns 1 if num is 0 or 1.
    """
    # Base cases: Factorial of 0 and 1 is 1
    if num == 0:
        return 1
    elif num == 1:
        return 1
    # Recursive case: Multiply num by the factorial of (num - 1)
    else:
        return num * factorial(num - 1)

if __name__ == "__main__":
    fact: int = factorial(5)
    print(fact)  # Output: 120
```

```python
def isprime(num):
    """
    Check if a number is prime.

    Args:
        num (int): The number to check for primality.
    Returns:
        bool: True if the number is prime, False otherwise. 
    """
    if num <= 1:
        return False
    else:
        for i in range(2, 9):
            if i == num:
             continue        
            elif num % i == 0:
                return False
        return True
```

```python
import random

def randomized_pass(length: int = 8) -> str:
    characters = "abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789"
    password =''
    """
    Generate a random password of specified length.

    Args:
        length (int): The desired length of the password. Default is 8.
    Returns:
        str: A randomly generated password consisting of letters and digits.
    """
    for _ in range(length):
        password += random.choice(characters)
    return password

random_password = randomized_pass(12)
print(random_password)  
```

```python
def repetition(n: str) -> dict:
    x=n.split()
    unique_words = set(x)
    word_count = {word: x.count(word) for word in unique_words}
    return word_count
print(repetition("hello world hello everyone i am adham hello"))
```

```python
import numpy as np

# 1. تعريف البيانات (مثال لبيانات وهمية لتجربة الكود)
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])

# 2. تجهيز المتغيرات والـ Hyperparameters
n = len(x)

w = 0
b = 0
learning_rate = 0.01
epochs = 100
cost_values = []

print(f"initial w : {w}")
print(f"initial b : {b}")
print(f"learning_rate : {learning_rate}")
print(f"epochs : {epochs}")
print("Training........................\n")

# 3. الـ Gradient Descent Loop
for i in range(epochs):
    # التوقع الأول
    y_hat = w * x + b
    
    # حساب الخطأ
    error = y_hat - y
    
    # حساب الـ Gradients
    d_w = np.mean(error * x)
    d_b = np.mean(error)
    
    # تحديث الـ Parameters
    w -= learning_rate * d_w
    b -= learning_rate * d_b
    
    # حساب التوقع والـ Cost الجديد
    y_hat = w * x + b
    cost = 0.5 * np.mean((y_hat - y) ** 2)
    cost_values.append(cost)
    
    # طباعة النتايج كل 100 إتريشن
    if (i + 1) % 100 == 0 or i == 0:
        print(f"Epoch {i+1}: cost = {cost:.4f}, w = {w:.4f}, b = {b:.4f}")
```
